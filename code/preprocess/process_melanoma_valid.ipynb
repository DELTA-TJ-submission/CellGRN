```r
library(DESeq2)

# library(data)
```

```r
# mel_count = read.table("/home/shaliu_fu/multireg/benchmark/datasets/melanoma/bulkRNAseq/GSE134432_bulkRNASeq_33MelanomaBaselines_raw_counts.tsv.gz",sep = "\t",header = T,row.names = 1)
egr3_kd = read.table("/home/shaliu_fu/multireg/benchmark/datasets/melanoma/bulkRNAseq/GSE134432_bulkRNASeq_EGR3_KD_raw_counts.tsv.gz",sep = "\t",header = T,row.names = 1)

nfatc2_kd = read.table("/home/shaliu_fu/multireg/benchmark/datasets/melanoma/bulkRNAseq/GSE134432_bulkRNASeq_NFATC2_KD_raw_counts.tsv.gz",sep = "\t",header = T,row.names = 1)
sox10_kd = read.table("/home/shaliu_fu/multireg/benchmark/datasets/melanoma/bulkRNAseq/GSE134432_bulkRNASeq_SOX10KD_raw_counts.tsv.gz",sep = "\t",header = T,row.names = 1)
```

```r
col_data = data.frame(row.names = colnames(egr3_kd),condition=factor(rep(c("ctl","kd"),2)))
dds <- DESeqDataSetFromMatrix(countData = as.matrix(egr3_kd), colData = col_data, design= ~ condition) 
dds <- DESeq(dds)
res <- results(dds, contrast=c("condition","kd","ctl")) #差异分析结果
res_out <- as.data.frame(res)
```

```r
egr3_kd_out<-subset(res_out,pvalue<0.1 & abs(log2FoldChange)>1)
```

```r
out_tab1 = data.frame(TF="EGR3",Gene=rownames(egr3_kd_out),Log2FC=egr3_kd_out$log2FoldChange,P_value=egr3_kd_out$pvalue)
```

```r
col_data = data.frame(row.names = colnames(nfatc2_kd),condition=factor(rep(c("ctl","kd"),2)))
dds <- DESeqDataSetFromMatrix(countData = as.matrix(nfatc2_kd), colData = col_data, design= ~ condition) #~在R里面用于构建公式对象，~左边为因变量，右边为自变量。
dds <- DESeq(dds)
res <- results(dds, contrast=c("condition","kd","ctl")) #差异分析结果
res_out <- as.data.frame(res)

```

```r
nfatc2_kd_out<-subset(res_out,pvalue<0.1 & abs(log2FoldChange)>1)
dim(nfatc2_kd_out)
```

```r
out_tab2 = data.frame(TF="NFATC2",Gene=rownames(nfatc2_kd_out),Log2FC=nfatc2_kd_out$log2FoldChange,P_value=nfatc2_kd_out$pvalue)
```

```r
col_data = data.frame(row.names = colnames(sox10_kd),condition=factor(rep(c("ctl","kd"),6)))
dds <- DESeqDataSetFromMatrix(countData = as.matrix(sox10_kd), colData = col_data, design= ~ condition) #~在R里面用于构建公式对象，~左边为因变量，右边为自变量。
dds <- DESeq(dds)
res <- results(dds, contrast=c("condition","kd","ctl")) #差异分析结果
res_out <- as.data.frame(res)
```

```r
sox10_kd_out<-subset(res_out,pvalue<0.01 & abs(log2FoldChange)>1)
dim(sox10_kd_out)
```

```r
out_tab3 = data.frame(TF="SOX10",Gene=rownames(sox10_kd_out),Log2FC=sox10_kd_out$log2FoldChange,P_value=sox10_kd_out$pvalue)
```

```r
out_tab<-rbind(out_tab1,out_tab2,out_tab3)
write.table(out_tab,file="/home/shaliu_fu/multireg/benchmark/datasets/melanoma/TF_gene_gold.txt",sep = "\t",row.names = F,col.names = T,quote = F)
```
