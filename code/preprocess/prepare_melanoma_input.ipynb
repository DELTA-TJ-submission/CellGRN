```python
import numpy as np
import pandas as pd
import os
import anndata as ad
from cellgrn.utils import load_scenic2, load_linger_all,load_figr, load_linger_ctx,generate_candidate_groups
```

```python
# 生成下每个算法对应的grn和input_genes和peaks
outdir = "/home/shaliu_fu/multireg/cellGRN/data/melanoma/"
os.system(f"mkdir -p {outdir}")
# dataset_config = json.load(open(f"/home/shaliu_fu/multireg/benchmark/bench_dataset/melanoma/RawData.json"))


gene_peak_res, tf_gene_res, tf_peak_res = load_scenic2('/home/shaliu_fu/multireg/benchmark/output/melanoma/', 'scenic2_output')



scenic_res = generate_candidate_groups([tf_gene_res,gene_peak_res,tf_peak_res])
out_genes = pd.DataFrame(list(set(tf_gene_res['Gene']) | set(tf_gene_res['TF']) | set(gene_peak_res['Gene']) | set(tf_peak_res['TF']) ))
out_peaks = pd.DataFrame(list(set(gene_peak_res['Peak']) | set(tf_peak_res['Peak']) ))


scenic_res.to_csv(f"{outdir}/scenic2_grn.csv",header=True,index=False)
out_genes.to_csv(f"{outdir}/scenic2_gene.txt",header=False,index=False)
out_peaks.to_csv(f"{outdir}/scenic2_peak.txt",header=False,index=False)
```
