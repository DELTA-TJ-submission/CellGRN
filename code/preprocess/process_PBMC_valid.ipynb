```r
# 按分组计算差异基因，fc, pvalue
get_pvalue <-function(mydata){
    return(wilcox.test(x=as.numeric(mydata[1:3]),
        y = as.numeric(mydata[4:6]),paired = T)$p.value)
}

myc_diff_tab<- data.frame(row.names = rownames(myc_tab),fc = rowMeans(myc_tab[,4:6]+0.01)/rowMeans(myc_tab[,1:3]+0.01),pvalue = apply(as.matrix(myc_tab),1,get_pvalue))
```

```r
get_pvalue <-function(mydata){
    return(c(wilcox.test(x=as.numeric(mydata[c(3,6,9,12)]),
        y = as.numeric(mydata[c(1,4,7,10)],paired=T))$p.value,
        wilcox.test(x=as.numeric(mydata[c(3,6,9,12)]),
        y = as.numeric(mydata[c(2,5,8,11)],paired=T))$p.value
        ))
}

foxp3_diff<- data.frame(row.names = rownames(foxp3_tab),
fc1 = rowMeans(foxp3_tab[,c(1,4,7,10)]+0.01)/rowMeans(foxp3_tab[,c(3,6,9,12)]+0.01),
fc2=rowMeans(foxp3_tab[,c(2,5,8,11)]+0.01)/rowMeans(foxp3_tab[,c(3,6,9,12)]+0.01))

foxp3_ptab<-data.frame(apply(as.matrix(foxp3_tab),1,get_pvalue))
```

```r
# dim(foxp3_ptab)
# dim(foxp3_diff)
foxp3_diff_tab<-cbind(foxp3_diff,t(foxp3_ptab))
```
